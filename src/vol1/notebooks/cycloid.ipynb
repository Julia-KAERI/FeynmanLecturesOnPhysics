# Cycloid 와 그 파생들

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Cycloid

```python
t = np.linspace(-np.pi/2, 3*np.pi/2, 1000)

plt.plot(np.cos(t), np.sin(t)+1)
ax = plt.gca()
ax.set_aspect("equal")

plt.plot(t-np.sin(t), (1-np.cos(t)))
plt.title("Cycloid")
plt.xlabel("$x$")
plt.ylabel("$y$")
```

## Curate Cycloid

```python
t = np.linspace(-np.pi/2, 3*np.pi/2, 1000)

plt.plot(np.cos(t), np.sin(t)+1)
ax = plt.gca()
ax.set_aspect("equal")

plt.plot((t-0.9*np.sin(t)), (1-0.9*np.cos(t)), label = "a=0.9")
plt.plot((t-0.8*np.sin(t)), (1-0.8*np.cos(t)), label = "a=0.8")
plt.plot((t-0.5*np.sin(t)), (1-0.5*np.cos(t)), label = "a=0.5")
plt.title("Curtate Cycloid")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.legend()
```

```python
τ0 = np.linspace(-np.pi/2, 3*np.pi/2, 1000)
def xx(a, τ):
    return a*np.sin(τ)

def zz(a, τ):
    return a*np.cos(τ)

R0=0

def xp(a, τ):
    t = τ + R0 + zz(a, τ)
    return (t, xx(a, τ))

plt.plot(xx(0.99, τ0), zz(0.99, τ0))
ax=plt.gca()
ax.set_aspect("equal")
```

```python
tt, xpr = xp(0.999, τ0)
plt.plot(tt, xpr)
```

```python
dtt = tt[1:]-tt[:-1]
tts = (tt[1:]+tt[:-1])*0.5
vxp = (xpr[1:] - xpr[:-1])/dtt
```

```python
plt.plot(tts, vxp)
```
