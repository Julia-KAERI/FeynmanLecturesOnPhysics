```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t=np.linspace(0, 30*np.pi, 300)
rectified = np.abs(np.sin(t*0.2))
plt.plot(t, rectified)
dt = t[1]-t[0]
```

```python
s_fft = np.fft.fft(rectified) # 추후 IFFT를 위해 abs를 취하지 않은 값을 저장한다.
amplitude = abs(s_fft)*(2/len(s_fft)) # 2/len(s)을 곱해줘서 원래의 amp를 구한다.
frequency = np.fft.fftfreq(len(s_fft), dt)
```

```python
plt.plot(frequency, amplitude)
```
