```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML
from matplotlib.animation import FuncAnimation

# 1. Figure와 Axes 생성
fig, ax = plt.subplots()
(line,) = ax.plot([], [], lw=2)
ax.set_xlim(0, 10 * np.pi)
ax.set_ylim(-2.5, 2.5)


# 2. 초기화 함수 (선택 사항)
def init():
  line.set_data([], [])

  return (line,)


# 3. 프레임마다 실행될 업데이트 함수
def update(frame):
  x = np.arange(0, frame, 0.2)
  y = np.cos(10*x + 5*frame ) + np.cos(5*np.sqrt(3)*x + frame/9.5*2)
  line.set_data(x, y)
  return (line,)


# 4. FuncAnimation 생성 (변수에 꼭 할당)
ani = FuncAnimation(
    fig, update, frames=100, init_func=init, interval=100, blit=True
)

HTML(ani.to_html5_video())
```
